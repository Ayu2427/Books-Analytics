```python
import pandas as pd
import matplotlib.pyplot as plt

# For the offline working sample
path = 'data/books_clean.csv'
df = pd.read_csv(path)
df.head()
```

## 1. Data Structure

```python
print(df.shape)
display(df.info())
print('\nMissing values:\n', df.isna().sum())
print('\nDuplicates:', df.duplicated().sum())
```

## 2. Meaningful Questions
- What is the typical book price?
- How are ratings distributed?
- Is price associated with rating?
- Which books are the most expensive?
- Are there price outliers?

```python
print(df[['Price_GBP','Rating_Num','Title_Length','Word_Count']].describe())
print('Price-rating correlation:', round(df['Price_GBP'].corr(df['Rating_Num']),3))
```

## 3. Visualizations

```python
plt.figure(figsize=(9,5))
plt.hist(df['Price_GBP'], bins=8)
plt.title('Book Price Distribution')
plt.xlabel('Price (£)')
plt.ylabel('Number of Books')
plt.show()
```

```python
counts=df['Rating'].value_counts().reindex(['One','Two','Three','Four','Five']).fillna(0)
plt.figure(figsize=(8,5))
plt.bar(counts.index, counts.values)
plt.title('Books by Rating')
plt.xlabel('Rating')
plt.ylabel('Number of Books')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['Rating_Num'], df['Price_GBP'])
plt.title('Price vs Rating')
plt.xlabel('Rating (1–5)')
plt.ylabel('Price (£)')
plt.show()
```

```python
top=df.nlargest(10,'Price_GBP').sort_values('Price_GBP')
plt.figure(figsize=(10,6))
plt.barh(top['Title'], top['Price_GBP'])
plt.title('Top 10 Most Expensive Books')
plt.xlabel('Price (£)')
plt.tight_layout()
plt.show()
```

## 4. Outlier Check

```python
q1,q3=df['Price_GBP'].quantile([.25,.75])
iqr=q3-q1
outliers=df[(df['Price_GBP']<q1-1.5*iqr)|(df['Price_GBP']>q3+1.5*iqr)]
outliers[['Title','Price_GBP','Rating']]
```
