# Books to Scrape — Web Scraping + EDA + Visualization
End-to-end notebook covering data collection, cleaning, exploratory analysis, and visualization.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('data/books_clean.csv')
df.head()
```

## Data Structure

```python
print(df.shape)
print(df.dtypes)
print('Missing values:', df.isna().sum())
print('Duplicates:', df.duplicated().sum())
```

## Descriptive Analysis

```python
print(df[['Price_GBP','Rating_Num','Title_Length','Word_Count']].describe())
print('Price-rating correlation:', round(df['Price_GBP'].corr(df['Rating_Num']),3))
```

## Visualizations

```python
plt.figure(figsize=(9,5))
plt.hist(df['Price_GBP'], bins=8)
plt.title('Book Price Distribution')
plt.xlabel('Price (£)')
plt.ylabel('Number of Books')
plt.show()
```

```python
counts=df['Rating'].value_counts().reindex(['One','Two','Three','Four','Five']).fillna(0)
plt.figure(figsize=(8,5))
plt.bar(counts.index, counts.values)
plt.title('Books by Rating')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['Rating_Num'], df['Price_GBP'])
plt.title('Price vs Rating')
plt.xlabel('Rating (1–5)')
plt.ylabel('Price (£)')
plt.show()
```

## ConclusionThe analysis summarizes price distribution, rating mix, price-rating association, and outlier checks. The source is a practice dataset.
